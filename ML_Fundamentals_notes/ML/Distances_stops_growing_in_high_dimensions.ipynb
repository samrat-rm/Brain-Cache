# Do distances stop growing in high dimensions?

Short answer: **no, the distances themselves never stop growing** — but the
*gap between your nearest and farthest neighbour* does, and that's the thing
worth remembering.

## What keeps growing

For `n` random points spread evenly inside a `d`-dimensional cube:

| Metric | Typical distance | At d = 2000 |
|---|---|---|
| Manhattan (L1) | grows like `d/3` | ~636 |
| Euclidean (L2) | grows like `sqrt(d/6)` | ~17.6 |

Both climb forever. Neither plateaus.

## What does stop growing

Measure `Dmax - Dmin`: how far apart the closest and the furthest point are
from your query. This behaves completely differently depending on the metric.

| Metric | Gap as d increases |
|---|---|
| Manhattan (L1) | keeps growing, like `sqrt(d)` |
| **Euclidean (L2)** | **converges to a constant (~1.35) and never moves again** |
| L3 and higher | actively shrinks toward zero |

The Euclidean gap is flat from about d = 20 onward. That's the result you were
half-remembering. It comes from Aggarwal, Hinneburg & Keim (2001),
*On the Surprising Behavior of Distance Metrics in High Dimensional Space* —
which is why they argued L1 beats L2 for high-dimensional nearest-neighbour
search, and fractional norms (L0.5) beat both.

## Why this matters in practice

Think of it like a photograph losing contrast. In low dimensions your nearest
neighbour is *clearly* closer than everything else. In high dimensions every
point ends up roughly the same distance away, so "nearest" becomes a coin flip
between near-identical numbers.

Look at the `rel` column above: at d = 10 the nearest point is ~3x closer than
the farthest. At d = 2000 the difference is under 10%. Below roughly 0.1, tiny
rounding errors decide your ranking instead of actual similarity.

Notice L1 and L2 have nearly the same `rel` column. **So switching from
Euclidean to Manhattan in a real search pipeline usually changes nothing
noticeable.** The theory favours L1, the practice barely does.

## Two things that surprise beginners

**1. Your 1536-dimensional embeddings are fine.**

The simulation above uses points scattered randomly everywhere. Real data
isn't like that — photos of cats, or sentence embeddings, cluster on a thin
curved sheet inside that huge space. A 1536-number embedding might really only
have ~20 dimensions worth of variation. Concentration follows that smaller
"real" number, not 1536. This is why vector search works at all despite the
maths looking hopeless.

**2. Most systems don't use raw Euclidean anyway.**

Vector databases usually rank by **cosine similarity** — the angle between two
vectors, ignoring their length. Since the distance-blowup problem is largely a
length problem, cosine sidesteps a good chunk of it. If you're using a vector
DB, you're probably already on cosine by default.

**Takeaway:** if retrieval quality is poor, don't swap L2 for L1. Get a better
embedding model, or reduce dimensions (PCA, a smaller model, Matryoshka
embeddings). Those actually move the needle.
